2. Muestra por pantalla:

La cantidad total de cada recurso.

El código de estado de cada petición.

```python
recursos = ['posts', 'comments', 'albums', 'photos', 'todos', 'users']

for recurso in recursos:
    url = f'https://jsonplaceholder.typicode.com/{recurso}'
    response = requests.get(url)

    if response.status_code == 200:
        datos = response.json()
        cantidad = len(datos)
        print(f"Recurso: {recurso:10} | Código: {response.status_code} | Total: {cantidad}")
    else:
        print(f"Recurso: {recurso:10} | Código: {response.status_code} | Error al obtener datos")
```

3. Realiza una petición a una publicación inexistente para obtener un error 404 y muestra el código de estado recibido.

```python
response = requests.get('https://jsonplaceholder.typicode.com/posts/9999')
print(f"Código de estado: {response.status_code}")
```

4. Realiza una petición POST para crear una nueva publicación ficticia. Incluye el título, el cuerpo del mensaje y un userId.


Muestra la respuesta JSON.

Muestra el código de estado.

```python
nueva_publicacion = {
    'title': 'Publicación de prueba',
    'body': 'Este es el contenido de mi publicación ficticia.',
    'userId': 1
}

response = requests.post('https://jsonplaceholder.typicode.com/posts', json=nueva_publicacion)

print(f"Código de estado: {response.status_code}")
print(f"Respuesta JSON: {response.json()}")
```

5. Realiza una petición PATCH para modificar parcialmente una publicación existente.


Muestra la respuesta JSON.

Muestra el código de estado.

```python
datos_actualizados = {
    'title': 'Título actualizado'
}

response = requests.patch('https://jsonplaceholder.typicode.com/posts/1', json=datos_actualizados)

print(f"Código de estado: {response.status_code}")
print(f"Respuesta JSON: {response.json()}")
```

6. Haz una petición DELETE sobre una publicación.

Muestra la respuesta JSON.

Muestra el código de estado.

```python
response = requests.delete('https://jsonplaceholder.typicode.com/posts/1')

print(f"Código de estado: {response.status_code}")
print(f"Respuesta JSON: {response.json()}")
```
